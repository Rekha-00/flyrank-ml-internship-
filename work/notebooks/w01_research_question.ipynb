## 3. Quick look at the data (2-3 real numbers)

I loaded the starter dataset and checked its size, the number of pages marked as declining by the starter proxy, and a couple of data-quality indicators. These numbers help show whether there is enough variation and coverage to investigate page prioritization.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Rekha-00/flyrank-ml-internship-/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

down_count = (df["trend_direction"] == "down").sum()
down_rate = down_count / len(df) * 100
missing_word_count = df["word_count"].isna().sum()

print(f"Number of pages: {len(df):,}")
print(f"Pages marked 'down' by the starter proxy: {down_count:,} ({down_rate:.1f}%)")
print(f"Pages with missing word_count: {missing_word_count:,}")
```
